```python
import pandas as pd

df = pd.read_csv('data/Indonesian_Food_Recipes.csv')
```

```python
import re

def clean_ingredients(text):
    text = text.lower()
    
    # hapus angka & satuan
    text = re.sub(r'\d+|\b(gr|kg|sdm|sdt|buah|butir|lembar|siung)\b', '', text)
    
    # ganti separator
    text = text.replace('--', ',')
    
    # hapus karakter aneh
    text = re.sub(r'[^a-zA-Z, ]', '', text)
    
    # split jadi list
    items = text.split(',')
    
    # bersihkan spasi
    items = [i.strip() for i in items if i.strip() != '']
    
    return items
```

```python
df['Ingredients Cleaned'] = df['Ingredients'].apply(clean_ingredients)
```

```python
stopwords = [
    'secukupnya', 'sesuai', 'selera', 'potong', 'dipotong',
    'cuci', 'dadu', 'iris', 'fillet', 'bagian', 'gunakan'
]

def clean_items(items):
    cleaned = []
    for item in items:
        words = item.split()
        
        # ambil 2 kata pertama saja (biar tidak jadi kalimat panjang)
        item_clean = ' '.join(words[:2])
        
        # buang stopwords
        if item_clean not in stopwords:
            cleaned.append(item_clean)
    
    return cleaned
```

```python
import re

recipe_stopwords = {
    "ikat", "secukupnya", "sesuai", "selera",
    "cincang", "halus", "kasar", "geprek",
    "memarkan", "iris", "tipis", "potong",
    "dadu", "serong", "lembar", "buah",
    "butir", "siung", "sendok", "sdt", "sdm",
    "gram", "kg", "ml", "liter", "cup",
    "gelas", "bungkus", "batang", "ruas",
    "ambil", "kupas", "cuci", "rebus",
    "goreng", "sangrai", "parut", "untuk",
    "tambahan", "taburan",

    # tambahan final cleaning
    "bks", "cm", "uk", "dan", "atau",
    "dengan", "bagian", "cairkan",
    "bubuk", "toping", "topping",
    "lalap", "kulit", "isi",
    "besar", "kecil", "sedang"
}

normalization = {
    "saos": "saus",
    "cabe": "cabai",
    "bamer": "bawang merah",
    "baput": "bawang putih"
}

def final_clean_v2(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    words = text.split()
    cleaned = []

    for word in words:
        # normalisasi
        word = normalization.get(word, word)

        # hapus stopwords
        if word not in recipe_stopwords:
            cleaned.append(word)

    # hapus duplikat
    cleaned = list(dict.fromkeys(cleaned))

    return " ".join(cleaned)

df["ingredients_final"] = df["ingredients_final"].apply(final_clean_v2)

print(df[["ingredients_final"]].head(10))
```

```python
import pandas as pd
import ast

# =========================
# LOAD DATA
# =========================
df = pd.read_csv("resep_cleaned.csv")

# cek kolom
print(df.columns)


# =========================
# UBAH STRING LIST → LIST ASLI
# karena Ingredients Cleaned masih berbentuk string
# =========================
def convert_to_list(x):
    try:
        return ast.literal_eval(x)
    except:
        return []

df["Ingredients Cleaned"] = df["Ingredients Cleaned"].apply(convert_to_list)


# =========================
# DAFTAR NOISE YANG HARUS DIHAPUS
# =========================
noise_words = {
    "ikat", "gelas", "air", "ekor", "bagian",
    "bungkus", "bks", "cm", "ruas", "batang",
    "lembar", "buah", "butir", "siung",
    "sendok", "sdm", "sdt", "kg", "gram",
    "ml", "liter", "cup", "secukupnya",
    "sesuai", "selera", "untuk", "tambahan",
    "taburan", "penyedap rasa", "penyedap",
    "rasa", "cincang", "halus", "kasar",
    "geprek", "memarkan", "iris", "tipis",
    "potong", "dadu", "serong"
}


# =========================
# FUNGSI CLEANING FINAL
# =========================
def clean_ingredients_list(ingredients):
    cleaned = []

    for item in ingredients:
        item = str(item).lower().strip()

        # hapus noise phrase penuh
        if item in noise_words:
            continue

        # hapus kata noise di dalam phrase
        words = item.split()
        words = [w for w in words if w not in noise_words]

        # gabungkan ulang
        item = " ".join(words).strip()

        # skip jika kosong
        if item == "":
            continue

        # skip jika masih terlalu pendek
        if len(item) <= 2:
            continue

        cleaned.append(item)

    # hapus duplikat
    cleaned = list(dict.fromkeys(cleaned))

    return cleaned


# =========================
# PROSES CLEANING
# =========================
df["Ingredients Final"] = df["Ingredients Cleaned"].apply(clean_ingredients_list)


# =========================
# BUAT KOLOM UNTUK TF-IDF
# =========================
df["ingredients_join"] = df["Ingredients Final"].apply(lambda x: " ".join(x))


# =========================
# HAPUS DUPLIKAT RESEP
# =========================
df = df.drop_duplicates(
    subset=["Title Cleaned", "ingredients_join"]
)


# =========================
# SIMPAN FILE FINAL
# =========================
df.to_csv("resep_final_ready.csv", index=False, encoding="utf-8-sig")

print("SELESAI")
print("Dataset final siap untuk TF-IDF")

print(df[["Title Cleaned", "Ingredients Final", "ingredients_join"]].head(10))
```

```python
import pandas as pd
import ast

# =========================
# LOAD DATA
# =========================
df = pd.read_csv("resep_cleaned.csv")

print(df.columns)


# =========================
# UBAH STRING LIST -> LIST
# =========================
def convert_to_list(x):
    try:
        return ast.literal_eval(x)
    except:
        return []

df["Ingredients Cleaned"] = df["Ingredients Cleaned"].apply(convert_to_list)


# =========================
# PHRASE YANG HARUS DIHAPUS
# (noise / bukan bahan utama)
# =========================
remove_phrases = {
    "aku beli",
    "bahan sambal",
    "penyedap rasa",
    "garam dan",
    "maizena cairkan",
    "tomat uk",
    "garam sck",
    "teh b",
    "mangkok makaroni",
    "sedikit jahe",
    "sedikit gula",
    "sedikit garam"
}


# =========================
# NORMALISASI PHRASE
# =========================
normalize_phrases = {
    "saos": "saus",
    "cabe": "cabai",
    "cabe ijo": "cabai hijau",
    "ijo": "hijau",
    "bputih": "bawang putih",
    "bmerah": "bawang merah",
    "btg sereh": "sereh",
    "btr jeruk": "jeruk",
    "grm": "garam",
    "daun": "",  # hanya jika berdiri sendiri
}


# =========================
# WORD NOISE KECIL
# =========================
small_noise = {
    "dan", "atau", "uk", "sck",
    "btg", "btr", "aku", "beli",
    "sedikit", "secukupnya",
    "bagian", "gelas", "ekor",
    "air", "buah", "butir",
    "siung", "sendok", "sdm", "sdt",
    "kg", "gram", "ml", "liter",
    "cup", "bungkus"
}


# =========================
# CLEANING FINAL
# =========================
def clean_phrase_level(items):
    result = []

    for item in items:
        item = str(item).lower().strip()

        # skip phrase noise penuh
        if item in remove_phrases:
            continue

        # normalisasi phrase
        for old, new in normalize_phrases.items():
            if old in item:
                item = item.replace(old, new)

        # split kata
        words = item.split()

        # hapus word noise
        words = [w for w in words if w not in small_noise]

        item = " ".join(words).strip()

        # skip kosong
        if item == "":
            continue

        # skip terlalu pendek
        if len(item) <= 2:
            continue

        result.append(item)

    # hapus duplikat
    result = list(dict.fromkeys(result))

    return result


# =========================
# PROSES CLEANING
# =========================
df["Ingredients Final"] = df["Ingredients Cleaned"].apply(clean_phrase_level)


# =========================
# KOLOM UNTUK TF-IDF
# =========================
df["ingredients_join"] = df["Ingredients Final"].apply(
    lambda x: " ".join(x)
)


# =========================
# HAPUS DUPLIKAT FINAL
# =========================
df = df.drop_duplicates(
    subset=["Title Cleaned", "ingredients_join"]
)


# =========================
# SIMPAN FILE FINAL
# =========================
df.to_csv(
    "resep_final_ready.csv",
    index=False,
    encoding="utf-8-sig"
)

print("SELESAI")
print("Dataset final sudah siap untuk TF-IDF")

print(df[
    ["Title Cleaned", "Ingredients Final", "ingredients_join"]
].head(20))
```

```python
import pandas as pd
import ast

# load data
df = pd.read_csv("resep_cleaned.csv")

# ubah string list jadi list asli
def convert_to_list(x):
    try:
        return ast.literal_eval(x)
    except:
        return []

df["Ingredients Cleaned"] = df["Ingredients Cleaned"].apply(convert_to_list)

# daftar bahan valid (whitelist)
valid_ingredients = {
    "ayam", "daging ayam", "fillet ayam", "ceker ayam",
    "bawang putih", "bawang merah", "bawang daun",
    "cabai rawit", "cabai merah", "cabai hijau",
    "garam", "gula", "minyak", "minyak goreng",
    "jahe", "kunyit", "lengkuas", "kemiri",
    "ketumbar", "merica", "serai", "sereh",
    "daun salam", "daun jeruk",
    "jeruk nipis", "tomat", "kol", "wortel",
    "tepung", "tepung ayam", "tepung tapioka",
    "tepung bumbu", "maizena",
    "telur", "kecap", "kecap manis",
    "saus", "saus sambal", "saus tiram",
    "kaldu", "jamur", "nasi", "makaroni"
}

# normalisasi kata
normalization = {
    "cabe": "cabai",
    "cabe ijo": "cabai hijau",
    "ijo": "hijau",
    "saos": "saus",
    "salam": "daun salam",
    "bputih": "bawang putih",
    "bmerah": "bawang merah",
    "minya": "minyak",
    "btg sereh": "sereh"
}

def clean_final(items):
    result = []

    for item in items:
        item = str(item).lower().strip()

        # normalisasi
        for old, new in normalization.items():
            if old in item:
                item = item.replace(old, new)

        # hanya simpan yang ada di whitelist
        if item in valid_ingredients:
            result.append(item)

    # hapus duplikat
    result = list(dict.fromkeys(result))
    return result

# proses final
df["Ingredients Final"] = df["Ingredients Cleaned"].apply(clean_final)

# gabung untuk TF-IDF
df["ingredients_join"] = df["Ingredients Final"].apply(
    lambda x: " ".join(x)
)

# hapus data kosong
df = df[df["ingredients_join"].str.strip() != ""]

# hapus duplikat
df = df.drop_duplicates(
    subset=["Title Cleaned", "ingredients_join"]
)

# simpan final
df.to_csv("resep_final_ready.csv", index=False, encoding="utf-8-sig")

print(df[["Title Cleaned", "Ingredients Final", "ingredients_join"]].head(20))
print("Dataset final benar-benar siap TF-IDF")
```

```python
phrase_map = {
    "bawang putih": "bawang_putih",
    "bawang merah": "bawang_merah",
    "cabai rawit": "cabai_rawit",
    "cabai merah": "cabai_merah",
    "cabai hijau": "cabai_hijau",
    "jeruk nipis": "jeruk_nipis",
    "saus tiram": "saus_tiram",
    "saus sambal": "saus_sambal",
    "kecap manis": "kecap_manis",
    "daun salam": "daun_salam",
    "daun jeruk": "daun_jeruk",
    "minyak goreng": "minyak_goreng",
    "tepung tapioka": "tepung_tapioka",
    "tepung bumbu": "tepung_bumbu",
    "fillet ayam": "fillet_ayam",
    "ceker ayam": "ceker_ayam",
    "daging ayam": "daging_ayam"
}

def apply_phrase_map(text):
    text = str(text).lower()

    for old, new in phrase_map.items():
        text = text.replace(old, new)

    return text

df["ingredients_join"] = df["ingredients_join"].apply(apply_phrase_map)

print(df["ingredients_join"].head(10))
```
